```python
def supernacci(n):
    supernacciNumbers = [1]

    for i in range(2, n+1):
        newNumber = (i-1)%3
        if (newNumber == 0):
            supernacciNumbers.append(supernacciNumbers[-1])
            print(supernacciNumbers)
        elif (newNumber == 1):
            supernacciNumbers.append(supernacciNumbers[-1])
            print(supernacciNumbers)
        elif(newNumber == 2):
            sumNumber = sum(supernacciNumbers[-2:])
            supernacciNumbers.append(sumNumber)
            print(supernacciNumbers)
        else:
            print("Something is terribly wrong xd")
    return
```
