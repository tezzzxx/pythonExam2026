```python
import math

#City data : [latitude, longitude, city size]
cities = {
'Bergen': [60.364098, 5.404422, 15],
'Oslo': [59.926959, 10.779454, 10],
'Trondheim': [63.433654, 10.390277, 5]
}

#POI data: (lantitude, longtidude) : [name of poi, rating]
poi = {(63.426806, 10.396712): ['Nidarosdommen', 4.5],
(59.926699, 10.701075): ['Vigelandsparken', 5],
(59.907704, 10.753163): ['Operahuset', 4],
(60.397561, 5.322833): ['Bryggen', 4.5],
(60.387639, 5.321623): ['Universitetmuseet', 5],
(60.342901, 5.336796): ['Gamlehaugen', 4]
}

def lat_km(lat_value):
    return lat_value * 111

def lon_km(lon_value):
    return lon_value * 68

def distance(lat1, lon1, lat2, lon2):
    d_lat = lat_km(lat1-lat2)
    d_lon = lon_km(lon1-lon2)
    return math.sqrt(d_lat**2 + d_lon**2)

def closestPOI(target_lat, target_lon):
    closest_name = None
    closest_dist = float("inf")

    for (p_lat, p_lon), (name, rating) in poi.items():
        d = distance(target_lat, target_lon, p_lat, p_lon)
        if d < closest_dist:
            closest_dist = d
            closest_name = name

    return closest_name

def closestCity(target_lat, target_lon):
    closest_name = None
    closest_dist = float("inf")

    for name, (c_lat, c_lon, c_size) in cities.items():
        d = distance(target_lat, target_lon, c_lat, c_lon)
        if d < closest_dist:
            closest_dist = d
            closest_name = name

    return closest_name

user_lat = float(input("Enter latitude :"))
user_lon = float(input("Enter longitude :"))

print("Closest city: ", closestCity(user_lat, user_lon))
print("Closest POI: ", closestPOI(user_lat, user_lon))
```

3B

```python
computed_boundaries = {}

for city_name, data in cities.items():
    lat = data[0]
    lon = data[1]
    size = data[2]

    north = round(lat + size, 6)
    south = round(lat - size, 6)
    east = round(lon + size, 6)
    west = round(lon - size, 6)

    computed_boundaries[city_name] = [north, south, east, west]

print("Computed city boundaries: ")
for city, boundaries in computed_boundaries.items():
    print(city, ": North:", boundaries[0], ", South:", boundaries[1], ", East:", boundaries[2], ", West:", boundaries[3])
```

3C

```python
city_pois = {}

for city, boundaries in computed_boundaries.items():
    north, south, east, west = boundaries
    city_pois[city] = []

    for coordinates, info in poi.items():
        poi_lat = coordinates[0]
        poi_lon = coordinates[1]
        poi_name = info[0]

        if south <= poi_lat <= north and west <= poi_lon <= east:
             city_pois[city].append(poi_name)

for city, p_list in city_pois.items():
    print(f"\nPoints of interest that you can find in {city}:")
    for poi_name in p_list:
        print(f"{poi_name}")
```

3D

```python
print("Average customer rating per city:")

for city, boundaries in computed_boundaries.items():
    north, south, east, west = boundaries

    total_rating = 0
    poi_count = 0

    for coordinates, info in poi.items():
        poi_lat = coordinates[0]
        poi_lon = coordinates[1]
        poi_rating = info[1]

        if south <= poi_lat <= north and west <= poi_lon <= east:
            total_rating += poi_rating
            poi_count += 1

    if poi_count > 0:
        average_rating = round(total_rating / poi_count, 2)
        print(f"{city}: {average_rating}")
    else:
        print(f"{city}: No POIs found")


```

```python
print("POIs sorted by their distance from city center")

for city_name, city_info in cities.items():
    city_lat = city_info[0]
    city_lon = city_info[1]

    north, south, east, west = computed_boundaries[city_name]

    city_distances = []

    for poi_coordinates, poi_info in poi.items():
        poi_lat = poi_coordinates[0]
        poi_lon = poi_coordinates[1]
        poi_name = poi_info[0]

        if south <= poi_lat <= north and west <= poi_lon <= east:
            d = distance(city_lat, city_lon, poi_lat, poi_lon)

            city_distances.append((d, poi_name))

    city_distances.sort()

    print(f"\n{city_name}")
    for dist, name in city_distances:
        print(f"{name} ({round(dist, 2)} km away from {city_name})")
```
