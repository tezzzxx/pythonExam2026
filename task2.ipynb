```python
import random as rnd
```

```python
def getDiceGuess():
    guess = ''

    while guess not in ('1','2','3','4'):
        print('Guess the number on a 4- sided dice! Enter a number between 1 and 4:')
        guess = input()

    return guess

def rollDice():
    dice = rnd.randint(1, 4) # count of dots on a 4-sided dice

    for i in range(4):
        guess = int(getDiceGuess())
        if dice == guess:
            print("Great job, tries used: ", i+1)
            break
        else:
            print("Wrong guess, guesses left: ", 3-i)
    return


```

```python
rollDice()
```
