```python
import random as rnd
```

```python
def getDiceGuess():
    guess = 0

    while guess not in ('1 ','2 ','3 ','4 '):
        print('Guess the number on a 4- sided dice! Enter a number between 1 and 4:')
        guess = input()

    return guess

def rollDice():
    dice = rnd.randint(1, 4) # count of dots on a 4-sided dice
    guess = getDiceGuess()
    if dice == guess:
        print('You are good!')
    else:
        print('Sorry! Try to guess again!')

    guess = getDiceGuess()
    if dice == guess:
        print('Congrats! You got it!')
    else:
        print('Sorry! Your third chance!')

    guess = getDiceGuess()
    if dice == guess:
        print('Congrats! You got it!')
    else:
        print('Sorry! You still can try!')

    guess = getDiceGuess()
    if dice == guess:
        print('Congrats! You got it!')
    else:
        print('Nope. You are really bad at this game.')

    return
```

```python
rollDice()
```
